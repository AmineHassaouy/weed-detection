```python
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from PIL import Image
 
CLASS_NAMES = ['Black-grass', 'Charlock', 'Cleavers',
               'Common Chickweed', 'Common wheat', 'Fat Hen',
               'Loose Silky-bent', 'Maize', 'Scentless Mayweed',
               'Shepherds Purse', 'Small-flowered Cranesbill', 'Sugar beet']
 
# Charger le modele entraine
model = load_model('weed_detection_model.h5')
 
# Charger et preparer votre image
img = Image.open(r'C:\Users\ahass\OneDrive\Desktop\notebooks\projects\detection des mauvaises herbes\data\v2-plant-seedlings-dataset\Black-grass\6.png').convert('RGB')
img = img.resize((224, 224))
img_array = preprocess_input(np.array(img, dtype=np.float32))
img_array = np.expand_dims(img_array, axis=0)  # Ajouter dimension batch
 
# Predire
prediction = model.predict(img_array)
classe = CLASS_NAMES[np.argmax(prediction)]
confiance = np.max(prediction) * 100
 
print(f"Plante detectee : {classe}")
print(f"Confiance : {confiance:.1f}%")
```
